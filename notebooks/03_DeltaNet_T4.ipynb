# Team 3: SOTA Baselines — Gated DeltaNet Implementation & Evaluation
**Pure-PyTorch Gated DeltaNet Layer: Associative Recurrence vs Static Linear Attention**

This notebook benchmarks Gated DeltaNet with data-dependent update ($eta$) and decay ($lpha$) gates against Linear Attention.

```python
import os, sys
import pandas as pd
import matplotlib.pyplot as plt

df_delta = pd.read_csv('../data/deltanet_comparison.csv')
df_delta
```

## 1. Associative Recall Comparison
Gated DeltaNet preserves $87.1\% - 96.5\%$ recall, overcoming Linear Attention's collapse.

```python
plt.figure(figsize=(9, 5))
for method, grp in df_delta.groupby('method'):
    plt.plot(grp.sequence_length_N, grp.associative_recall_acc, marker='o', lw=2, label=method)
plt.xscale('log')
plt.xlabel('Sequence Length N')
plt.ylabel('Associative Recall Accuracy')
plt.title('Associative Recall: Gated DeltaNet vs Linear Attention vs SDPA')
plt.grid(True, alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
```

## 2. Throughput & Latency Scaling

```python
plt.figure(figsize=(9, 5))
for method, grp in df_delta.groupby('method'):
    plt.plot(grp.sequence_length_N, grp.latency_ms, marker='s', lw=2, label=method)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Sequence Length N')
plt.ylabel('Latency (ms, log scale)')
plt.title('Latency Scaling: DeltaNet vs Linear Attention')
plt.grid(True, alpha=0.4)
plt.legend()
plt.tight_layout()
plt.show()
```
